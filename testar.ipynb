```python
from pydantic_orders.models import Order

order = Order(
    order_id=1,
    customer_id=1001,
    product_category="Electronics",
    quantity=2,
    unit_price=499.99,
    discount=0.1,
)

print(order)    #Testar så att modelen fungerar 
```

```python
order2 = Order(
    order_id=1,
    customer_id=1001,
    product_category="Electronics",
    quantity= -2,
    unit_price=499.99,
    discount=0.1,
)

print(order2)    #### Testar fel order med negativ quantity för att få validation error
```

```python
from pydantic import BaseModel, Field, ValidationError

class User(BaseModel):
    age: int = Field(gt=0)



# Inte required när default finns

    value: float = Field(ge=0)                # Lägger till params i fälten som greater than och greater/equal 0


user = User(age=25, value=10.5)
print(user)

user2 = User(age = "28", value = "34.5")    #Typkonvertering till int och float från str
print(user2)


print(type(user.age))
print(type(user2.age))    #Testar visar nya typer


user.model_dump()

user2.model_dump_json()


User.model_fields
```
