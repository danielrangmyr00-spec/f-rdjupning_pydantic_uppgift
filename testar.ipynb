```python
from pydantic_orders.models import Order

order = Order(
    order_id=1,
    customer_id=1001,
    product_category="Electronics",
    quantity=2,
    unit_price=499.99,
    discount=0.1,
)

print(order)    #Testar så att modelen fungerar 
```

```python
order2 = Order(
    order_id=1,
    customer_id=1001,
    product_category="Electronics",
    quantity= -2,
    unit_price=499.99,
    discount=0.1,
)

print(order2)    #### Testar fel order med negativ quantity för att få validation error
```
